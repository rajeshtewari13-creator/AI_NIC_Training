Exercise 3: Generative AI — Prompt Engineering for Video Synthesis


## 1. Selected AI Video Generation Tool
**Tool Name:** OpenArt.ai  
**Mode:** Frame to Video / Image to Video  
**Website:** https://openart.ai/

## 2. Objective
To create a 3-second video demonstrating image blending using the two original images and the blended image generated in Exercise 2.

## 3. Input Images
- Image1.jpg – Mountain lake landscape.
- Image2.jpg – Yellow sunflower photograph.
- Exercise2_Blended.jpg – Combined image created using 60% Image 1 and 40% Image 2.

## 4. Exact Prompt Text

Create a professional 3-second cinematic educational video demonstrating AI image blending using the three supplied reference images.

Scene 1: Begin with the beautiful mountain lake landscape featuring snow-covered mountains, autumn trees and clear water reflections.

Scene 2: Smoothly transition to a bright yellow sunflower against a natural green background. Highlight the flower with a subtle camera zoom.

Scene 3: Transform the scene into a beautiful double-exposure composition, combining the sunflower and mountain lake landscape. Show both images visibly blended together while preserving their original colors, textures and details.

Use realistic visuals, smooth transitions, natural lighting and professional cinematic quality. Maintain visual consistency throughout the video. Do not introduce unrelated objects, people, text, logos or watermarks.

## 5. Camera and Transition Instructions

- 0–1 Second: Display Image1.jpg with a slow camera push-in.
- 1–2 Seconds: Transition smoothly to Image2.jpg using a cross-dissolve effect.
- 2–3 Seconds: Transition to Exercise2_Blended.jpg and hold the final blended composition.

Camera movement: Smooth and stable.
Transition: Cross-dissolve.
Aspect Ratio: 16:9 (Landscape).
Target Duration: 3 seconds.
Audio: Not required.

## 6. OpenArt.ai Implementation Steps

1. Open https://openart.ai/.
2. Navigate to Video Generation.
3. Select Frame to Video or Image to Video.
4. Choose a suitable video-generation model.
5. Upload the reference images using the available image or keyframe controls.
6. Enter the exact prompt provided above.
7. Set the landscape aspect ratio (16:9).
8. Generate the video.
9. If the selected model generates a longer video, trim it to exactly 3 seconds.
10. Review and save the final video.

If only start and end frames are supported, generate separate shots or use a storyboard/timeline workflow to incorporate all three images.

## 7. Expected Output

A 3-second cinematic video showing:
- Original mountain landscape.
- Original sunflower photograph.
- Final blended image demonstrating the image-blending effect.

## 8. Conclusion

This exercise demonstrates the application of prompt engineering in multimodal Generative AI. OpenArt.ai is used to create a short video by combining image references, structured scene descriptions, camera instructions and transition effects.
