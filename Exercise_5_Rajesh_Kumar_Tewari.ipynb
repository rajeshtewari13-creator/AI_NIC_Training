Exercise 5: Interactive AI Prototyping with Gradio

```python
import os
import gradio as gr

# Optional: set OPENAI_API_KEY as an environment variable before running.
# Example (Windows PowerShell): $env:OPENAI_API_KEY="your_api_key"
# Do not write or share your secret API key in the notebook submitted publicly.

def chatbot_response(message, history):
    api_key = os.getenv("OPENAI_API_KEY")

    if api_key:
        try:
            from openai import OpenAI
            client = OpenAI(api_key=api_key)

            messages = [
                {
                    "role": "system",
                    "content": "You are a helpful, clear, and professional assistant. Answer the user's questions concisely."
                }
            ]

            # Include recent conversation context
            for item in history or []:
                if isinstance(item, dict):
                    if item.get("role") in ("user", "assistant"):
                        messages.append({"role": item["role"], "content": item.get("content", "")})
                elif isinstance(item, (list, tuple)) and len(item) == 2:
                    user_text, assistant_text = item
                    if user_text:
                        messages.append({"role": "user", "content": user_text})
                    if assistant_text:
                        messages.append({"role": "assistant", "content": assistant_text})

            messages.append({"role": "user", "content": message})

            response = client.chat.completions.create(
                model="gpt-4o-mini",
                messages=messages,
                temperature=0.7
            )
            return response.choices[0].message.content

        except Exception as error:
            return f"AI service could not be reached. Please check the API key and connection. Details: {error}"

    # Demonstration response when no API key is configured
    return (
        "Welcome to the GOT-A demonstration chatbot! "
        "You asked: " + message + "\n\n"
        "This is a demonstration response. To enable AI-generated answers, "
        "configure the OPENAI_API_KEY environment variable and run this cell again."
    )

demo = gr.ChatInterface(
    fn=chatbot_response,
    title="GOT-A AI Assistant Chatbot",
    description=(
        "Welcome! Ask a question and the assistant will respond. "
        "Try one of the sample prompts below. Configure an OpenAI API key "
        "to enable live AI-generated responses."
    ),
    examples=[
        "What is Artificial Intelligence?",
        "Explain the difference between AI and Machine Learning."
    ]
)

demo.launch()
```
