```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Reproducible random values
np.random.seed(42)

# 1. Create a DataFrame with four numerical columns
df = pd.DataFrame({
    "f1": np.random.randint(1, 100, 10),
    "f2": np.random.randint(1, 100, 10),
    "f3": np.random.randint(1, 100, 10),
    "f4": np.random.randint(1, 100, 10)
})

print("Original DataFrame:")
display(df)

# 2. Scatter plot showing the relationship between f2 and f3
plt.figure(figsize=(8, 5))
plt.scatter(df["f2"], df["f3"], marker="o")
plt.xlabel("f2 values")
plt.ylabel("f3 values")
plt.title("Scatter Plot: Relationship between f2 and f3")
plt.grid(True, alpha=0.3)
plt.show()

# 3. Calculate row-wise arithmetic mean across f1, f2, f3 and f4
df["f5"] = df[["f1", "f2", "f3", "f4"]].mean(axis=1)

print("DataFrame after adding f5 (row-wise mean):")
display(df)

# 4. Export all five columns to an Excel file
output_excel = "Exercise1_Data.xlsx"
df.to_excel(output_excel, index=False)

print(f"Excel file saved successfully: {output_excel}")
```
