Exercise 2: Computer Vision — Image Blending Using OpenCV

```python
import cv2
import matplotlib.pyplot as plt
from pathlib import Path

image1_path = "image1.jpg"   # Change this path if required
image2_path = "image2.jpg"   # Change this path if required

# Load images
img1 = cv2.imread(image1_path)
img2 = cv2.imread(image2_path)

if img1 is None:
    raise FileNotFoundError(f"Could not load {image1_path}. Check the file path and image format.")
if img2 is None:
    raise FileNotFoundError(f"Could not load {image2_path}. Check the file path and image format.")

# Resize second image to the dimensions of the first image
height, width = img1.shape[:2]
img2_resized = cv2.resize(img2, (width, height))

# Weighted blending: 60% image 1 + 40% image 2
blended = cv2.addWeighted(img1, 0.6, img2_resized, 0.4, 0)

# Save all three images
cv2.imwrite("Exercise2_Image1.jpg", img1)
cv2.imwrite("Exercise2_Image2.jpg", img2_resized)
cv2.imwrite("Exercise2_Blended.jpg", blended)

# Convert BGR (OpenCV format) to RGB for correct display in Matplotlib
img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2_resized, cv2.COLOR_BGR2RGB)
blended_rgb = cv2.cvtColor(blended, cv2.COLOR_BGR2RGB)

# Display original and blended images side by side
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
axes[0].imshow(img1_rgb)
axes[0].set_title("Original Image 1")
axes[1].imshow(img2_rgb)
axes[1].set_title("Original Image 2 (Resized)")
axes[2].imshow(blended_rgb)
axes[2].set_title("Blended Image")
for ax in axes:
    ax.axis("off")
plt.tight_layout()
plt.show()

print("Saved: Exercise2_Image1.jpg, Exercise2_Image2.jpg, Exercise2_Blended.jpg")
```
